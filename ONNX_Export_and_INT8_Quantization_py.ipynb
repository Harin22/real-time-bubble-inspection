```python
from onnxruntime.quantization import quantize_dynamic, QuantType

#Quantize the ONNX model to INT8
quantize_dynamic(
    model_input='/content/drive/MyDrive/Real-Time Bubble [ PIPELINE ]/best_segmentation0.1v.onnx',
    model_output='/content/drive/MyDrive/Real-Time Bubble [ PIPELINE ]/best_segmentation0.1v_int8.onnx',
    weight_type=QuantType.QUInt8  #Used 8-bit integers
)

print(" Quantized to INT8: best_segmentation0.1v_int8.onnx")
```
